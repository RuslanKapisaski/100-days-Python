```python
import pandas as pd

# Adjust display format
pd.options.display.float_format = '{:,.2f}'.format
```

```python
df = pd.read_csv("salaries_by_college_major.csv")
df.head()
```

```python
# Check the labels of the columns
df.columns
```

```python
# Check for NaN values
df.isna
```

```python
# Peek the end of the data frame
df.tail()
```

```python
# Create a clean data frame without NULL values from the original data frame
clean_df = df.dropna()
```

```python
clean_df.tail()
```

```python
# Find College Major with Highest Starting Salaries
starting_median_salaries = clean_df['Starting Median Salary']
highest_staring_median_salaries = starting_median_salaries.max()
highest_staring_median_salaries
```

```python
# Find the index of the college with max salary
college_idx = starting_median_salaries.idxmax()
college_idx
```

```python
# Find the name of the major that corresponds to the max salary
major = clean_df['Undergraduate Major'].loc[43]
# major = clean_df['Undergraduate Major'][43]
major
```

```python
# What college major has the highest mid-career salary?
print(clean_df['Mid-Career Median Salary'].max())
print(f"Index for the max mid career salary: {clean_df['Mid-Career Median Salary'].idxmax()}")
clean_df['Undergraduate Major'][8]
```

```python
# What college major has the lowest mid-career salary?
min_salary = clean_df['Mid-Career Median Salary'].min()
print(f"Min mid career salary: {min_salary}")
min_mid_salary_idx = clean_df['Mid-Career Median Salary'].idxmax()
print(f"Index for the min mid career salary: {min_mid_salary_idx}")
clean_df['Undergraduate Major'][8]
```

```python
# Calculate the difference between the earnings of the 10th and 90th percentile
spread_col = clean_df['Mid-Career 90th Percentile Salary'] - clean_df['Mid-Career 10th Percentile Salary']
clean_df.insert(1, 'Spread', spread_col)
clean_df.head()
```

```python
# See which degrees have the smallest spread
lowest_sorted = clean_df.sort_values(by=['Spread'])
lowest_sorted.head()
```

```python
# Find the top 5 degrees with the highest values in the 90th percentile
highest_sorted = clean_df.sort_values(by=['Spread'], ascending=False)
highest_sorted.head()
```

```python
# Which majors have the largest difference between high and low earners after graduation
highest_spread = clean_df.sort_values('Spread', ascending=False)
highest_spread[["Undergraduate Major","Spread", "Group"]].head()
```

```python
# Count how many majors we have in each category
clean_df.groupby("Group").count()
```

```python
# Find the average spread of the majors
clean_df. groupby('Group')["Spread"].mean()
```
